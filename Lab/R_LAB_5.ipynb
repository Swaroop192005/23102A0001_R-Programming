```r
# ==============================================================================
# LAB 5 - SOCIAL NETWORK ANALYSIS (SNA) WITH R
# Term & Document Co-occurrence Network from Twitter data (apple.csv)
# COMPLETE SCRIPT - runs end to end in Google Colab (R runtime)
# ------------------------------------------------------------------------------
# BEFORE RUNNING: click the folder icon in the left sidebar ->
# "Upload to session storage" -> select apple.csv
# ==============================================================================

# ------------------------------------------------------------------------------
# Step 0: Colab environment setup (system libs + fast binary CRAN mirror)
# ------------------------------------------------------------------------------
cat("R version:", R.version.string, "\n")
options(repr.plot.width = 10, repr.plot.height = 7, repr.plot.res = 120)

system("apt-get update -qq", ignore.stdout = TRUE, ignore.stderr = TRUE)
system(paste("apt-get install -y -qq",
             "libglpk-dev libxml2-dev libgmp-dev",
             "libcurl4-openssl-dev libssl-dev"),
       ignore.stdout = TRUE, ignore.stderr = TRUE)

options(HTTPUserAgent = sprintf(
  "R/%s R (%s)", getRversion(),
  paste(getRversion(), R.version$platform, R.version$arch, R.version$os)))
ubuntu_codename <- tryCatch(system("lsb_release -cs", intern = TRUE),
                            error = function(e) "jammy")
options(repos = c(CRAN = paste0("https://packagemanager.posit.co/cran/__linux__/",
                                ubuntu_codename, "/latest")))

# ------------------------------------------------------------------------------
# Step 1: Install & Load Required Packages
# ------------------------------------------------------------------------------
required_packages <- c("tm", "igraph", "RColorBrewer")
new_packages <- required_packages[!(required_packages %in% installed.packages()[, "Package"])]
if (length(new_packages)) {
  cat("Installing:", paste(new_packages, collapse = ", "), "\n")
  tryCatch(install.packages(new_packages, quiet = TRUE),
           error = function(e) install.packages(new_packages,
                                                repos = "https://cloud.r-project.org",
                                                quiet = TRUE))
}
suppressPackageStartupMessages({
  library(tm)
  library(igraph)
  library(RColorBrewer)
})
for (p in required_packages) cat(sprintf("  %-14s %s\n", p, as.character(packageVersion(p))))

# ------------------------------------------------------------------------------
# Step 2: Read and Prepare Data   (file.choose() replaced - no GUI in Colab)
# ------------------------------------------------------------------------------
csv_url    <- ""    # optional public raw URL to apple.csv
candidates <- c("apple.csv", "/content/apple.csv", "/content/sample_data/apple.csv")
csv_path   <- candidates[file.exists(candidates)][1]
if (is.na(csv_path) && nzchar(csv_url)) {
  csv_path <- "/content/apple.csv"; download.file(csv_url, csv_path, quiet = TRUE)
}
if (is.na(csv_path) || !file.exists(csv_path))
  stop("apple.csv not found. Upload it via the Files sidebar, then re-run.")

apple <- read.csv(csv_path, header = TRUE, stringsAsFactors = FALSE)
cat("\nLoaded:", csv_path, "|", nrow(apple), "rows x", ncol(apple), "columns\n")
if (!"text" %in% names(apple))
  stop("No 'text' column. Found: ", paste(names(apple), collapse = ", "))
apple$text[is.na(apple$text)] <- ""

# Build Corpus using universal UTF-8 encoding (Windows / Mac / Linux safe)
corpus <- iconv(apple$text, to = "UTF-8", sub = "byte")
corpus[is.na(corpus)] <- ""
corpus <- Corpus(VectorSource(corpus))
cat("Corpus built with", length(corpus), "documents.\n")

# ------------------------------------------------------------------------------
# Step 3: Text Cleaning (compatible with modern 'tm')
# ------------------------------------------------------------------------------
removeURL <- function(x) gsub("http[[:alnum:]]*", "", x)

corpus   <- tm_map(corpus, content_transformer(tolower))
corpus   <- tm_map(corpus, content_transformer(removePunctuation))
corpus   <- tm_map(corpus, content_transformer(removeNumbers))
cleanset <- tm_map(corpus,   removeWords, stopwords("english"))
cleanset <- tm_map(cleanset, content_transformer(removeURL))
cleanset <- tm_map(cleanset, removeWords, c("aapl", "apple"))

replace_word <- content_transformer(function(x, pattern, replacement) {
  return(gsub(pattern, replacement, x))
})
cleanset <- tm_map(cleanset, replace_word, pattern = "stocks", replacement = "stock")
cleanset <- tm_map(cleanset, stripWhitespace)

# ------------------------------------------------------------------------------
# Step 4: Build Term-Document Matrix (TDM)
# ------------------------------------------------------------------------------
tdm   <- TermDocumentMatrix(cleanset)
tdm_m <- as.matrix(tdm)
cat("Full TDM:", nrow(tdm_m), "terms x", ncol(tdm_m), "documents\n")

frequent_rows <- rowSums(tdm_m) > 30
if (sum(frequent_rows) == 0) {
  warning("No terms exceeded threshold > 30. Lowering threshold to > 5.")
  tdm_m <- tdm_m[rowSums(tdm_m) > 5, ]
} else {
  tdm_m <- tdm_m[frequent_rows, ]
}
cat("Retained", nrow(tdm_m), "frequent terms.\n\nTerm frequencies:\n")
print(sort(rowSums(tdm_m), decreasing = TRUE))

cat("\nTop 10 terms and first 10 documents matrix:\n")
print(tdm_m[1:min(10, nrow(tdm_m)), 1:min(10, ncol(tdm_m))])

# ------------------------------------------------------------------------------
# Step 5: Network of Terms (Co-occurrence Network)
# ------------------------------------------------------------------------------
tdm_binary <- tdm_m
tdm_binary[tdm_binary > 1] <- 1

termM  <- tdm_binary %*% t(tdm_binary)
g_term <- graph_from_adjacency_matrix(termM, weighted = TRUE,
                                      mode = "undirected", diag = FALSE)
g_term <- simplify(g_term)
V(g_term)$label  <- V(g_term)$name
V(g_term)$degree <- degree(g_term)
cat("\nTerm network:", vcount(g_term), "nodes,", ecount(g_term), "edges\n")

options(repr.plot.width = 9, repr.plot.height = 6)
hist(V(g_term)$degree, breaks = 20, col = "limegreen",
     main = "Histogram of Term Node Degrees",
     ylab = "Frequency", xlab = "Degree of Vertices (Connections)")

options(repr.plot.width = 11, repr.plot.height = 9)
set.seed(222)
plot(g_term, vertex.color = "lightgreen", vertex.size = 6,
     vertex.label.dist = 1.2, main = "Term Co-occurrence Network")

# ------------------------------------------------------------------------------
# Step 6: Community Detection Algorithms
# ------------------------------------------------------------------------------
options(repr.plot.width = 18, repr.plot.height = 6)
par(mfrow = c(1, 3)); set.seed(222)

comm_eb <- cluster_edge_betweenness(g_term)
plot(comm_eb, g_term, main = "Edge Betweenness")

comm_lp <- cluster_label_prop(g_term)
plot(comm_lp, g_term, main = "Label Propagation")

comm_fg <- cluster_fast_greedy(g_term)
plot(comm_fg, g_term, main = "Fast Greedy")

par(mfrow = c(1, 1))

cat("\n--- Community detection summary ---\n")
cat(sprintf("%-20s : %d communities, modularity = %.4f\n",
            "Edge Betweenness", length(comm_eb), modularity(comm_eb)))
cat(sprintf("%-20s : %d communities, modularity = %.4f\n",
            "Label Propagation", length(comm_lp), modularity(comm_lp)))
cat(sprintf("%-20s : %d communities, modularity = %.4f\n",
            "Fast Greedy", length(comm_fg), modularity(comm_fg)))
cat("\nFast Greedy community membership:\n")
for (k in sort(unique(membership(comm_fg))))
  cat("  Community", k, ":",
      paste(V(g_term)$name[membership(comm_fg) == k], collapse = ", "), "\n")

# ------------------------------------------------------------------------------
# Step 7: Centrality Analysis (Hubs & Authorities)
# ------------------------------------------------------------------------------
# Works on igraph 1.x and 2.x (hub_score/authority_score deprecated in 2.x)
get_hits <- function(g) {
  r <- tryCatch(suppressWarnings(list(
         hub       = igraph::hub_score(g,       weights = NA)$vector,
         authority = igraph::authority_score(g, weights = NA)$vector)),
       error = function(e) NULL)
  if (is.null(r))
    r <- tryCatch({ h <- igraph::hits_scores(g, weights = NA)
                    list(hub = h$hub, authority = h$authority) },
                  error = function(e) NULL)
  if (is.null(r) || is.null(r$hub) || is.null(r$authority)) {
    ev <- igraph::eigen_centrality(g, weights = NA)$vector
    r  <- list(hub = ev, authority = ev)
  }
  r
}
hits <- get_hits(g_term); hs <- hits$hub; auth <- hits$authority

options(repr.plot.width = 16, repr.plot.height = 8)
par(mfrow = c(1, 2)); set.seed(222)
plot(g_term, vertex.size = hs * 30, main = "Hubs (Term Importance)",
     vertex.label = V(g_term)$name, vertex.label.cex = 0.8,
     vertex.color = rainbow(length(V(g_term))))
plot(g_term, vertex.size = auth * 30, main = "Authorities (Term Centrality)",
     vertex.label = V(g_term)$name, vertex.label.cex = 0.8,
     vertex.color = rainbow(length(V(g_term))))
par(mfrow = c(1, 1))

cat("\nTop 10 hub scores:\n");       print(round(sort(hs,   decreasing = TRUE)[1:min(10, length(hs))],   4))
cat("\nTop 10 authority scores:\n"); print(round(sort(auth, decreasing = TRUE)[1:min(10, length(auth))], 4))

# ------------------------------------------------------------------------------
# Step 8: Highlighting High-Degree Term Nodes
# ------------------------------------------------------------------------------
max_deg <- max(V(g_term)$degree)
V(g_term)$label.cex   <- (1.5 * V(g_term)$degree / max_deg) + 0.5
V(g_term)$label.color <- rgb(0, 0, 0.3, 0.9)
V(g_term)$frame.color <- NA

edge_weights <- E(g_term)$weight
if (is.null(edge_weights)) edge_weights <- rep(1, ecount(g_term))
egam <- (log(edge_weights + 1)) / max(log(edge_weights + 1))
E(g_term)$color <- rgb(0.5, 0.5, 0, egam)
E(g_term)$width <- egam * 2

options(repr.plot.width = 11, repr.plot.height = 9); set.seed(222)
plot(g_term, vertex.color = "lightgreen", vertex.size = V(g_term)$degree * 0.8,
     main = "Weighted Term Co-occurrence Network")

# ------------------------------------------------------------------------------
# Step 8b: Quantitative network measures (Tasks 7 & 8)
# ------------------------------------------------------------------------------
cat("\n=== Graph-level measures (term network) ===\n")
cat(sprintf("Nodes (terms)            : %d\n",   vcount(g_term)))
cat(sprintf("Edges                    : %d\n",   ecount(g_term)))
cat(sprintf("Density                  : %.4f\n", edge_density(g_term)))
cat(sprintf("Average degree           : %.2f\n", mean(degree(g_term))))
cat(sprintf("Transitivity (clustering): %.4f\n", transitivity(g_term)))
cat(sprintf("Diameter (unweighted)    : %d\n",   as.integer(diameter(g_term, weights = NA))))
cat(sprintf("Mean distance            : %.4f\n", mean_distance(g_term)))
cat(sprintf("Connected components     : %d\n",   components(g_term)$no))

cent <- data.frame(
  term            = V(g_term)$name,
  degree          = degree(g_term),
  weighted_degree = strength(g_term),
  betweenness     = round(betweenness(g_term, weights = NA), 2),
  closeness       = round(closeness(g_term,   weights = NA), 5),
  eigenvector     = round(eigen_centrality(g_term, weights = NA)$vector, 4),
  hub             = round(hs,   4),
  authority       = round(auth, 4),
  row.names = NULL, stringsAsFactors = FALSE)
cent <- cent[order(-cent$degree, -cent$betweenness), ]

cat("\n=== Term centrality table (ranked by degree) ===\n")
print(head(cent, 20), row.names = FALSE)
cat("\nMost influential terms:", paste(head(cent$term, 5), collapse = ", "), "\n")

# ------------------------------------------------------------------------------
# Step 9: Network of Documents / Tweets
# ------------------------------------------------------------------------------
tweetM  <- t(tdm_binary) %*% tdm_binary
g_tweet <- graph_from_adjacency_matrix(tweetM, weighted = TRUE,
                                       mode = "undirected", diag = FALSE)
g_tweet <- simplify(g_tweet)
V(g_tweet)$degree <- degree(g_tweet)
cat("\nTweet network:", vcount(g_tweet), "nodes,", ecount(g_tweet), "edges\n")

options(repr.plot.width = 9, repr.plot.height = 6)
hist(V(g_tweet)$degree, breaks = 20, col = "lightblue",
     main = "Histogram of Tweet Node Degrees", ylab = "Frequency", xlab = "Degree")

options(repr.plot.width = 11, repr.plot.height = 9); set.seed(222)
plot(g_tweet, vertex.label = NA, vertex.size = 3, vertex.color = "skyblue",
     main = "Tweet Similarity Network")

# Filter Low-Degree Nodes (prune weakly connected tweets)
degree_threshold <- quantile(V(g_tweet)$degree, 0.75)   # keep top 25% connected
g2 <- delete_vertices(g_tweet, V(g_tweet)[degree(g_tweet) < degree_threshold])
cat("Degree threshold (75th pct):", round(degree_threshold),
    "| Filtered network:", vcount(g2), "nodes,", ecount(g2), "edges\n")

set.seed(222)
plot(g2, vertex.size = 4, vertex.label = NA, vertex.color = "coral",
     main = paste("Filtered Tweet Network (Degree >=", round(degree_threshold), ")"))

# ------------------------------------------------------------------------------
# Step 10: Interpretation of Results
# ------------------------------------------------------------------------------
top5      <- head(cent, 5)
bt        <- cent[order(-cent$betweenness), ]
iso_terms <- V(g_term)$name[degree(g_term) == 0]

cat("\n==========================================================\n")
cat("  INTERPRETATION OF THE SOCIAL NETWORK ANALYSIS\n")
cat("==========================================================\n\n")

cat("1. NETWORK CONSTRUCTION\n")
cat(sprintf("   Term network: %d nodes, %d edges, density %.3f. A node is a word;\n",
            vcount(g_term), ecount(g_term), edge_density(g_term)))
cat("   an edge means the two words were tweeted together at least once.\n")
cat(sprintf("   Document network: %d tweet-nodes, %d shared-term links.\n\n",
            vcount(g_tweet), ecount(g_tweet)))

cat("2. INFLUENTIAL / HIGHLY CONNECTED NODES\n")
for (i in seq_len(nrow(top5)))
  cat(sprintf("     %d. %-12s degree = %3d | betweenness = %8.2f | eigenvector = %.3f\n",
              i, top5$term[i], top5$degree[i], top5$betweenness[i], top5$eigenvector[i]))
cat(sprintf("   '%s' is the most connected term and acts as the hub of the conversation.\n\n",
            top5$term[1]))

cat("3. BRIDGING NODES\n")
cat(sprintf("   Highest betweenness: %s (%.2f) - lies on the shortest paths between\n",
            bt$term[1], bt$betweenness[1]))
cat("   other terms and therefore connects otherwise separate sub-topics.\n\n")

cat("4. COMMUNITY STRUCTURE\n")
cat(sprintf("   Fast Greedy: %d communities (modularity %.3f); Edge Betweenness: %d (%.3f);\n",
            length(comm_fg), modularity(comm_fg), length(comm_eb), modularity(comm_eb)))
cat(sprintf("   Label Propagation: %d (%.3f).\n", length(comm_lp), modularity(comm_lp)))
if (modularity(comm_fg) < 0.3) {
  cat("   Low modularity => one densely interconnected conversation rather than\n   clearly separated discussion groups (typical of a single trending topic).\n\n")
} else {
  cat("   The modularity indicates reasonably well-separated topic clusters.\n\n")
}

cat("5. DEGREE DISTRIBUTION\n")
cat(sprintf("   Term degrees range %d to %d (mean %.1f). The right-skewed shape means a\n",
            as.integer(min(degree(g_term))), as.integer(max(degree(g_term))),
            mean(degree(g_term))))
cat("   few terms carry most connections - the scale-free pattern of social networks.\n")
if (length(iso_terms)) cat("   Isolated terms:", paste(iso_terms, collapse = ", "), "\n")
cat("\n")

cat("6. DOCUMENT NETWORK\n")
cat(sprintf("   After pruning at the 75th percentile (degree >= %d), %d of %d tweets remain -\n",
            as.integer(round(degree_threshold)), vcount(g2), vcount(g_tweet)))
cat("   the dense core of near-duplicate / thematically similar messages (retweets).\n")
cat("\n==========================================================\n")

# ------------------------------------------------------------------------------
# Save all figures and result tables for submission
# ------------------------------------------------------------------------------
dir.create("outputs", showWarnings = FALSE)
save_png <- function(file, expr, width = 1200, height = 900) {
  png(file.path("outputs", file), width = width, height = height, res = 120)
  on.exit(dev.off(), add = TRUE)
  expr
  invisible(file)
}
set.seed(222)
save_png("01_term_degree_histogram.png",
         hist(V(g_term)$degree, breaks = 20, col = "limegreen",
              main = "Histogram of Term Node Degrees",
              ylab = "Frequency", xlab = "Degree of Vertices (Connections)"))
save_png("02_term_network.png",
         plot(g_term, vertex.color = "lightgreen", vertex.size = 6,
              vertex.label.dist = 1.2, main = "Term Co-occurrence Network"))
save_png("03_communities.png", {
  par(mfrow = c(1, 3))
  plot(comm_eb, g_term, main = "Edge Betweenness")
  plot(comm_lp, g_term, main = "Label Propagation")
  plot(comm_fg, g_term, main = "Fast Greedy")
  par(mfrow = c(1, 1))
}, width = 2100, height = 750)
save_png("04_hubs_authorities.png", {
  par(mfrow = c(1, 2))
  plot(g_term, vertex.size = hs * 30, main = "Hubs (Term Importance)",
       vertex.label = V(g_term)$name, vertex.label.cex = 0.8,
       vertex.color = rainbow(length(V(g_term))))
  plot(g_term, vertex.size = auth * 30, main = "Authorities (Term Centrality)",
       vertex.label = V(g_term)$name, vertex.label.cex = 0.8,
       vertex.color = rainbow(length(V(g_term))))
  par(mfrow = c(1, 1))
}, width = 1900, height = 900)
save_png("05_weighted_term_network.png",
         plot(g_term, vertex.color = "lightgreen",
              vertex.size = V(g_term)$degree * 0.8,
              main = "Weighted Term Co-occurrence Network"))
save_png("06_tweet_degree_histogram.png",
         hist(V(g_tweet)$degree, breaks = 20, col = "lightblue",
              main = "Histogram of Tweet Node Degrees",
              ylab = "Frequency", xlab = "Degree"))
save_png("07_tweet_network.png",
         plot(g_tweet, vertex.label = NA, vertex.size = 3,
              vertex.color = "skyblue", main = "Tweet Similarity Network"))
save_png("08_filtered_tweet_network.png",
         plot(g2, vertex.size = 4, vertex.label = NA, vertex.color = "coral",
              main = paste("Filtered Tweet Network (Degree >=",
                           round(degree_threshold), ")")))

write.csv(cent, "outputs/term_centrality_measures.csv", row.names = FALSE)
write.csv(data.frame(term = V(g_term)$name,
                     fast_greedy_community = as.integer(membership(comm_fg))),
          "outputs/term_communities.csv", row.names = FALSE)
write.csv(data.frame(term = rownames(tdm_m), frequency = as.integer(rowSums(tdm_m))),
          "outputs/term_frequencies.csv", row.names = FALSE)
system("zip -qr sna_outputs.zip outputs")

cat("\nSaved files:\n"); print(list.files("outputs"))
cat("\nArchive: sna_outputs.zip (download from the Files sidebar)\n")
cat("\nScript executed successfully!\n")
```
