```r
###############################################################
# Assignment 1 - R Programming
# Task 1: Import and Inspect the Dataset
###############################################################

# Install required package (only if not already installed)
if(!require(dplyr)){
  install.packages("dplyr", repos = "https://cloud.r-project.org")
  library(dplyr)
}

# -------------------------------
# Specify the CSV file name
# -------------------------------
file_name <- "PRSA_Data_Aotizhongxin_20130301-20170228.csv"

# -------------------------------
# Read the dataset using tryCatch()
# -------------------------------
air_data <- tryCatch({

  read.csv(file_name)

}, warning = function(w){

  cat("Warning:", conditionMessage(w), "\n")

}, error = function(e){

  cat("Error:", conditionMessage(e), "\n")
  NULL

})

# -------------------------------
# Stop if dataset not loaded
# -------------------------------
if(is.null(air_data)){
  stop("Dataset could not be loaded.")
}

###############################################################
# Display first six records
###############################################################

cat("\n==============================\n")
cat("First 6 Records\n")
cat("==============================\n")

head(air_data)

###############################################################
# Display structure
###############################################################

cat("\n==============================\n")
cat("Dataset Structure\n")
cat("==============================\n")

str(air_data)

###############################################################
# Number of rows and columns
###############################################################

cat("\n==============================\n")
cat("Dataset Dimensions\n")
cat("==============================\n")

cat("Rows    :", nrow(air_data), "\n")
cat("Columns :", ncol(air_data), "\n")

###############################################################
# Check missing values
###############################################################

cat("\n==============================\n")
cat("Missing Values Check\n")
cat("==============================\n")

contains_missing <- any(is.na(air_data))

cat("Contains Missing Values :", contains_missing, "\n")

total_missing <- sum(is.na(air_data))

cat("Total Missing Values :", total_missing, "\n")

###############################################################
# Summary Statistics
###############################################################

cat("\n==============================\n")
cat("Summary Statistics\n")
cat("==============================\n")

summary(air_data)

###############################################################
# Column Names
###############################################################

cat("\n==============================\n")
cat("Column Names\n")
cat("==============================\n")

print(colnames(air_data))
```

```r
###############################################################
# Assignment 1 - R Programming
# Task 2: Understand NA, NULL and NaN
# Task 3: Missing Value Summary Function
###############################################################

###############################################################
# TASK 2
###############################################################

cat("\n=====================================================\n")
cat("TASK 2 : NA, NULL and NaN Demonstration\n")
cat("=====================================================\n")

# Example of NA
temperature <- c(28, 30, NA, 32)

# Example of NULL
missing_object <- NULL

# Example of NaN
undefined_value <- 0 / 0

cat("\nTemperature Vector:\n")
print(temperature)

cat("\nNULL Object:\n")
print(missing_object)

cat("\nNaN Value:\n")
print(undefined_value)

cat("\nChecking NA:\n")
print(is.na(temperature))

cat("\nChecking NULL:\n")
print(is.null(missing_object))

cat("\nChecking NaN:\n")
print(is.nan(undefined_value))


###############################################################
# TASK 3
###############################################################

cat("\n=====================================================\n")
cat("TASK 3 : Missing Value Summary\n")
cat("=====================================================\n")

missing_summary <- function(df){

  selected_variables <- c(
    "PM2.5",
    "PM10",
    "SO2",
    "NO2",
    "TEMP",
    "WSPM",
    "wd"
  )

  summary_table <- data.frame(
    Variable = character(),
    Total_Records = integer(),
    Missing_Values = integer(),
    Missing_Percentage = numeric(),
    stringsAsFactors = FALSE
  )

  for(variable in selected_variables){

    if(variable %in% names(df)){

      total_records <- nrow(df)

      missing_values <- sum(is.na(df[[variable]]))

      missing_percentage <- round(
        (missing_values / total_records) * 100,
        2
      )

      summary_table <- rbind(
        summary_table,
        data.frame(
          Variable = variable,
          Total_Records = total_records,
          Missing_Values = missing_values,
          Missing_Percentage = missing_percentage
        )
      )

      if(missing_percentage > 20){

        warning(
          paste(
            variable,
            "contains more than 20% missing values."
          )
        )

      }

    }

  }

  return(summary_table)

}

###############################################################
# Generate Missing Summary
###############################################################

missing_summary_table <- missing_summary(air_data)

cat("\nMissing Value Summary:\n")

print(missing_summary_table)

###############################################################
# Total Missing Values Per Column
###############################################################

cat("\nMissing Values in Every Column:\n")

print(colSums(is.na(air_data)))
```

```r
###############################################################
# Assignment 1 - R Programming
# Task 4: Identify Invalid Numerical Results
# Task 5: Handle Missing Numerical Values Using Loop
###############################################################

###############################################################
# TASK 4 : pollution_ratio
###############################################################

cat("\n=====================================================\n")
cat("TASK 4 : Pollution Ratio\n")
cat("=====================================================\n")

# Create pollution_ratio
air_data$pollution_ratio <- air_data$PM2.5 / air_data$PM10

# Count invalid values
na_count <- sum(is.na(air_data$pollution_ratio))
nan_count <- sum(is.nan(air_data$pollution_ratio))
inf_count <- sum(is.infinite(air_data$pollution_ratio))

cat("NA Count       :", na_count, "\n")
cat("NaN Count      :", nan_count, "\n")
cat("Infinite Count :", inf_count, "\n")

# Replace NaN and Infinite values with NA
air_data$pollution_ratio[
  is.nan(air_data$pollution_ratio) |
  is.infinite(air_data$pollution_ratio)
] <- NA

cat("\nInvalid values successfully replaced with NA.\n")


###############################################################
# TASK 5 : Handle Missing Numerical Values
###############################################################

cat("\n=====================================================\n")
cat("TASK 5 : Numerical Missing Value Treatment\n")
cat("=====================================================\n")

numeric_variables <- c(
  "PM2.5",
  "PM10",
  "SO2",
  "NO2",
  "TEMP",
  "WSPM"
)

# Store missing values before and after cleaning
before_missing <- c()
after_missing <- c()
median_used <- c()

for(variable in numeric_variables){

  cat("\n--------------------------------------\n")
  cat("Variable :", variable, "\n")

  # Check whether column exists
  if(!(variable %in% names(air_data))){
    cat("Column does not exist.\n")
    next
  }

  # Missing values before cleaning
  before <- sum(is.na(air_data[[variable]]))

  # Median ignoring NA
  med <- median(air_data[[variable]], na.rm = TRUE)

  # Replace NA with median
  air_data[[variable]][
    is.na(air_data[[variable]])
  ] <- med

  # Missing values after cleaning
  after <- sum(is.na(air_data[[variable]]))

  cat("Missing Before :", before, "\n")
  cat("Median Used    :", med, "\n")
  cat("Missing After  :", after, "\n")

  before_missing <- c(before_missing, before)
  after_missing  <- c(after_missing, after)
  median_used    <- c(median_used, med)

}

###############################################################
# Summary Table
###############################################################

numeric_cleaning_summary <- data.frame(

  Variable = numeric_variables,

  Missing_Before = before_missing,

  Median_Used = median_used,

  Missing_After = after_missing

)

cat("\n=====================================================\n")
cat("Summary of Numerical Cleaning\n")
cat("=====================================================\n")

print(numeric_cleaning_summary)
```

```r
###############################################################
# Assignment 1 - R Programming
# Task 6: Handle Missing Categorical Values
# Task 7: Error Handling using tryCatch()
###############################################################

###############################################################
# TASK 6 : Handle Missing Values in wd
###############################################################

cat("\n=====================================================\n")
cat("TASK 6 : Handling Missing Values in 'wd'\n")
cat("=====================================================\n")

# Function to calculate mode
calculate_mode <- function(x){

  # Remove missing values
  x <- x[!is.na(x)]

  # Unique values
  unique_values <- unique(x)

  # Most frequent value
  mode_value <- unique_values[
    which.max(tabulate(match(x, unique_values)))
  ]

  return(mode_value)
}

# Missing values before replacement
before_wd <- sum(is.na(air_data$wd))

# Calculate mode
mode_wd <- calculate_mode(air_data$wd)

cat("Mode of wd :", mode_wd, "\n")

# Replace missing values
air_data$wd[is.na(air_data$wd)] <- mode_wd

# Missing values after replacement
after_wd <- sum(is.na(air_data$wd))

cat("Missing Before :", before_wd, "\n")
cat("Missing After  :", after_wd, "\n")


###############################################################
# TASK 7 : Error Handling Function
###############################################################

cat("\n=====================================================\n")
cat("TASK 7 : clean_variable() using tryCatch()\n")
cat("=====================================================\n")

clean_variable <- function(data, variable){

  tryCatch({

    # Variable exists?
    if(!(variable %in% names(data))){
      stop("Variable does not exist.")
    }

    # Variable numeric?
    if(!is.numeric(data[[variable]])){
      stop("Categorical variable passed instead of numerical.")
    }

    # Entire column missing?
    if(all(is.na(data[[variable]]))){
      stop("Variable contains only missing values.")
    }

    # Calculate median
    med <- median(data[[variable]], na.rm = TRUE)

    # Median valid?
    if(is.na(med)){
      stop("Median cannot be calculated.")
    }

    # Replace missing values
    data[[variable]][
      is.na(data[[variable]])
    ] <- med

    cat(variable, "cleaned successfully.\n")

    return(data[[variable]])

  },

  error = function(e){

    cat("ERROR :", conditionMessage(e), "\n")

    return(NULL)

  })

}

###############################################################
# Demonstration of clean_variable()
###############################################################

cat("\nCleaning PM2.5\n")
clean_variable(air_data, "PM2.5")

cat("\nTrying categorical variable (wd)\n")
clean_variable(air_data, "wd")

cat("\nTrying non-existing variable\n")
clean_variable(air_data, "ABCXYZ")
```

```r
###############################################################
# Assignment 1 - R Programming
# Task 8: Comparison Table
# Task 9: Missing Value Visualization
###############################################################

###############################################################
# TASK 8 : Comparison Table
###############################################################

cat("\n=====================================================\n")
cat("TASK 8 : Comparison Table\n")
cat("=====================================================\n")

comparison_table <- data.frame(

  Variable = c(numeric_variables, "wd"),

  Missing_Before = c(before_missing, before_wd),

  Missing_After = c(after_missing, after_wd),

  Values_Replaced = c(before_missing, before_wd) -
                    c(after_missing, after_wd)

)

print(comparison_table)

###############################################################
# Interpretation
###############################################################

cat("\n=====================================================\n")
cat("Interpretation\n")
cat("=====================================================\n")

if(all(comparison_table$Missing_After == 0)){

  cat("All selected missing values have been handled successfully.\n")

}else{

  cat("Some missing values are still present.\n")

}


###############################################################
# TASK 9 : Bar Chart
###############################################################

cat("\n=====================================================\n")
cat("TASK 9 : Bar Chart\n")
cat("=====================================================\n")

# Create matrix for plotting
missing_matrix <- rbind(

  comparison_table$Missing_Before,

  comparison_table$Missing_After

)

# Plot bar chart
barplot(

  missing_matrix,

  beside = TRUE,

  col = c("red", "green"),

  names.arg = comparison_table$Variable,

  main = "Missing Values Before and After Cleaning",

  xlab = "Variables",

  ylab = "Number of Missing Values",

  ylim = c(0, max(comparison_table$Missing_Before) + 20)

)

legend(

  "topright",

  legend = c("Before Cleaning", "After Cleaning"),

  fill = c("red", "green")

)

###############################################################
# Save Graph
###############################################################

png("Missing_Values_Bar_Chart.png",
    width = 900,
    height = 600)

barplot(

  missing_matrix,

  beside = TRUE,

  col = c("red", "green"),

  names.arg = comparison_table$Variable,

  main = "Missing Values Before and After Cleaning",

  xlab = "Variables",

  ylab = "Number of Missing Values",

  ylim = c(0, max(comparison_table$Missing_Before) + 20)

)

legend(

  "topright",

  legend = c("Before Cleaning", "After Cleaning"),

  fill = c("red", "green")

)

dev.off()

cat("\nGraph saved as 'Missing_Values_Bar_Chart.png'\n")
```
