This cell initializes a variable `x` with `NA` (Not Available) to demonstrate how to check for missing values in R using `is.na()`.

```r
# Missing data in R

x <- NA

# Check whether x is missing
is.na(x)
```

This cell creates a vector `x` containing numeric values and `NA`s, then uses `is.na()` to identify the positions of missing values.

```r
x <- c(11, NA, 13, NA)

x

# Check which values are NA
is.na(x)
```

This cell calculates the mean of a vector `x` with missing values, showing the default behavior (returning `NA`) and how to calculate the mean after removing `NA`s using `na.rm = TRUE`.

```r
x <- c(11, NA, 13, NA)

# Mean including NA
mean(x)

# Mean after removing NA
mean(x, na.rm = TRUE)
```

This cell differentiates between `NA` and `NULL` in R. `NA` represents a missing value, while `NULL` represents an empty object. It demonstrates checking for each using `is.na()` and `is.null()` respectively.

```r
x <- NA
y <- NULL

print(x)
print(y)

# Check whether they are NA
is.na(x)
is.null(y)
```

This cell counts the total number of `NA` values in a vector `x` by summing the logical results of `is.na(x)` (where `TRUE` is 1 and `FALSE` is 0).

This cell uses `complete.cases()` to return a logical vector indicating which elements of `x` are *not* missing (i.e., complete).

This cell demonstrates `na.omit()`, which returns a new vector `y` with all `NA` values removed from the original vector `x`.

This cell demonstrates a simple `if` statement in R. If `x` is greater than 4, it performs a calculation, otherwise, it does nothing.

```r
x <- 5

if (x > 4) {
  x * 3
}
```

This cell demonstrates an `if` statement that prints a message if the condition `x > 3` is true.

```r
x <- 6

if (x > 3) {
  print("The value is more than 3")
}
```

This cell illustrates an `if-else` statement. If `x` equals 3, it subtracts 1 from `x`; otherwise, it multiplies `x` by 2.

```r
x <- 5

if (x == 3) {
  x <- x - 1
} else {
  x <- 2 * x
}

x
```

This cell demonstrates a basic `if-else` statement, printing different messages based on whether `x` is greater than 3.

```r
x <- 6

if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}
```

This cell shows an `if-else` statement where the `else` block is executed because the condition `x > 3` is false.

```r
x <- 2

if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}
```

This cell demonstrates an `if-else if-else` structure. It checks multiple conditions sequentially and executes the corresponding code block.

```r
x <- 5

if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}

x
```

This cell is another example of `if-else if-else`, where the `else if` condition (`x < 3`) is met, and `x` is updated accordingly.

```r
x <- 2

if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}

x
```

This cell demonstrates an `if-else if-else` structure where the first `if` condition (`x == 3`) is met, and `x` is updated.

```r
x <- 3

if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}

x
```

This cell introduces the `ifelse()` function, which is a vectorized version of the `if-else` statement. It applies a condition to each element of a vector `x` and returns a new vector based on the true or false outcome for each element.

```r
x <- 1:10

ifelse(x < 6, x^2, x + 1)
```

This cell is a repeat of the previous `ifelse()` example, showing its behavior for a vector `x` from 1 to 10.

```r
x <- 1:10

ifelse(x < 6, x^2, x + 1)
```

This cell demonstrates the `switch()` function with a numeric index. It selects the Nth argument based on the value of the first argument (2 in this case), returning 'banana'.

```r
switch(2, "apple", "banana", "orange")
```

This cell demonstrates the `switch()` function with a string as the first argument. It matches the string 'colour' to the named argument `colour = "blue"` and returns 'blue'.

```r
switch(
  "colour",
  colour = "blue",
  gender = "male",
  volume = 50
)
```

This cell shows two examples of `switch()`: one with a numeric index that doesn't match any argument (returning `NULL`), and another with a string index 'size' that also doesn't match any named argument, resulting in `NULL`.

```r
switch(4, "apple", "banana", "orange")
switch(
  "size",
  colour = "blue",
  gender = "male",
  volume = 50
)
```

This cell creates the same 3x3 matrix `x` and then uses `which()` to find the linear indices of elements that are odd (i.e., have a remainder of 1 when divided by 2).

```r
x <- matrix(
  nrow = 3,
  ncol = 3,
  data = 1:9
)

x

which(x %% 2 == 1)
```

```r
which(x %% 2 == 1, arr.ind = TRUE)
```
