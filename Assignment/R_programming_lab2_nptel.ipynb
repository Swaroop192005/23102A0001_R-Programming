This cell demonstrates vector recycling in R: when adding vectors of different lengths, the shorter vector is recycled to match the length of the longer vector. Here, the length of `y` (2) is a factor of the length of `x` (4).

```r
x <- c(24, 20, 8, 16)
y <- c(3, 4, 2, 8)

x / y
```

This cell shows an example of vector recycling where the length of the shorter vector (`y`, length 3) is not a multiple of the longer vector (`x`, length 4), which results in a warning, but the operation still proceeds by recycling `y` partially.

```r
x <- c(2, 3, 5, 7)
y <- c(8, 9)

x + y
```
