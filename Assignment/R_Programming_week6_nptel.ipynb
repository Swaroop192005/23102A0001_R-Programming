This cell demonstrates a basic `for` loop in R, iterating from 1 to 5 and printing the square of each number.

```r
x <- c(2, 4, 6, 7)

for (i in x) {
  print(i^2)
}
```

This cell shows a `for` loop iterating over the elements of a numeric vector `x` and printing the square of each element.

```r
x <- c(2, 4, 6, 8, 10, 12)

count <- 0

for (xval in x) {
  if (xval / 2 > 3) {
    count <- count + 1
  }
}

print(count)
```

This cell demonstrates a `for` loop with an `if` condition. It iterates through a vector `x` and increments a `count` variable if an element divided by 2 is greater than 3, finally printing the total count.

```r
child <- c("child1", "child2", "child3")
sweet <- c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

This cell illustrates a nested `for` loop. It iterates through two vectors, `child` and `sweet`, printing all possible combinations of elements from both vectors.

```r
drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }

  print(x)
}
```

This cell demonstrates the use of the `break` statement within a `for` loop. It iterates through a vector `drink` and stops printing elements once 'tea' is encountered.

```r
drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }

  print(x)
}
```

This cell demonstrates the use of the `next` statement within a `for` loop. It iterates through a vector `drink`, skipping 'lemonade' and printing the other elements.

```r
i <- 1

while (i < 10) {
  print(i^2)
  i <- i + 2
}
```

This cell shows a `while` loop that continues as long as `i` is less than 10. It prints the square of `i` and then increments `i` by 2 in each iteration.

```r
i <- 1

repeat {
  print(i^2)
  i <- i + 2

  if (i > 10) {
    break
  }
}
```

This cell demonstrates a `repeat` loop, which is an infinite loop that continues until a `break` condition is met. It prints the square of `i`, increments `i` by 2, and breaks when `i` exceeds 10.

This cell defines a simple R function `abc` that takes one argument `x` and returns its square. It then calls this function with different values.

```r
abc <- function(x, y) {
  x^2 + y^2
}

abc(3, 4)
abc(10, 10)
abc(-2, -3)
```
